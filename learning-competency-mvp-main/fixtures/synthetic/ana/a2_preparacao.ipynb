# Preparação e exploração dos dados
Dados sintéticos: `atendimentos_2026.csv` (exportação fictícia do sistema de chamados).

## 1. Carregar dados

```
import pandas as pd
df = pd.read_csv('dados/atendimentos_2026.csv', parse_dates=['abertura'])
print(df.shape)
```

## 2. Limpeza

```
# remover registros sem duração e duplicados
df = df.dropna(subset=['duracao_min']).drop_duplicates('id')
print(len(df))
```

## 3. Exploração inicial

```
df.groupby('mes')['duracao_min'].mean().round(1)
```
