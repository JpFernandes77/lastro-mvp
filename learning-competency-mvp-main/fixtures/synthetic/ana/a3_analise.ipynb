# Análise reproduzível — fatores associados ao TMA

## 1. TMA por canal, turno e tipo

```
import pandas as pd
df = pd.read_csv('dados/atendimentos_2026_limpo.csv')
tab = df.pivot_table(index='canal', columns='mes', values='duracao_min', aggfunc='mean').round(1)
print(tab)
```

## 2. Participação de solicitações de 2º nível

```
# share de chamados escalados por mês
df.groupby('mes')['escalado'].mean().round(3)
```

## 3. Modelo simples

```
import statsmodels.formula.api as smf
m = smf.ols('duracao_min ~ C(canal) + C(turno) + escalado', data=df).fit()
print(m.params.round(2))
```
